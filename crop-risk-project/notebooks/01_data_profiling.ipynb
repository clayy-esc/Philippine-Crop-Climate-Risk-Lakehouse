```python
from pyspark.sql import functions as F

crop_path = "/Volumes/crop_risk/source_data/raw/crop_trend_master.csv"
weather_path = "/Volumes/crop_risk/source_data/raw/province_weather_quarterly.csv"

df_crop = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(crop_path)
)

df_weather = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(weather_path)
)
```

```python
df_crop.select(
    F.min("year"),
    F.max("year")
).show()
```

```python
df_crop.groupBy("quarter").count().show()
```

```python
missing = df_crop.select([
    F.sum(F.col(c).isNull().cast("int")).alias(c)
    for c in df_crop.columns
])

display(missing)
```

```python
business_key = [
    "province",
    "crop",
    "year",
    "quarter"
]

df_crop.groupBy(business_key).count().filter(F.col("count") > 1).display()
```

```python
duplicates = (
    df_crop
    .groupBy(business_key)
    .count()
    .filter(F.col("count") > 1)
)

display(duplicates)
```
